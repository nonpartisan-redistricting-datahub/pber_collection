```python
def get_race(race_string):
    race_string = race_string.title()
    race_string = race_string.replace("(Vote For 1)","")
    if "U.S. House" in race_string or 'Us House' in race_string or "Representative To Congress" in race_string or "U. S. Representative" in race_string or "United States Representative" in race_string or "Congress" in race_string:
        return "CON"
    elif "State House" in race_string or "State Representative" in race_string or "Representative" in race_string:
        return "SL"
    elif "State Senate" in race_string or "State Senator" in race_string or "Senate" in race_string:
        return "SU"
    elif "President" in race_string:
        return "PRE"
    elif "US Senate" in race_string or "Us Senate" in race_string or "U.S. Senator" in race_string or "U. S. Senator" in race_string or "United States Senator" in race_string:
        return "USS"
    elif "Public Service" in race_string:
        return "PSC"
    elif "Attorney General" in race_string:
        return "ATG"
    elif "Auditor General" in race_string or "Auditor" in race_string:
        return "AUD"
    elif "Treasurer" in race_string:
        return "TRE"
    elif "Superintendent" in race_string:
        return "SUP"
    elif "Secretary Of State" in race_string:
        return "SOS"
    elif "Governor" in race_string:
        return "GOV"
    elif "Lieutenant Governor" in race_string:
        return "LTG"

    elif "Commissioner Of Labor" in race_string:
        return "LAB"
    elif "Commissioner Of Agriculture" in race_string:
        return "AGR"
    elif "Commissioner Of Insurance" in race_string:
        return "INS"
    elif "Chief Justice Of The Supreme Court" in race_string:
        return "CJU"
    elif "Supreme" in race_string:
        return "SC"
    elif "Justice Of The Supreme Court" in race_string:
        return "JUS"
    elif "Amendment" in race_string:
        #num = race_string.split(" ")[2][:-1]
        return "AMD"
    elif "Bond Question" in race_string:
        num = race_string.split(" ")[2][:-1]
        return "BQ"+str(num)
    elif "Judicial Retention" in race_string:
        return "JR"
    elif "Commissioner Of Public Lands" in race_string:
        return "CPL"
    elif "Comptroller" in race_string:
        return "COM"
    
    elif "All Medically Appropriate Assisted" in race_string:
        return "BI1"
    
    elif "Should Any Candidate Appearing On The Illinois Ballot For Federal" in race_string:
        return "BI3"
    
    elif "Illinois Constitution Be Amended To Create An Additional 3%" in race_string:
        return "BI2"

    else:
        print("No race for:", race_string)
        raise ValueError
        
def get_election_type(race_string):
    return "G"
        
def get_party(race_string):
    race_string = race_string.replace('"',"'")

    if "JUDICIAL RETENTION" in race_string or "Amendment" in race_string or "amendment" in race_string or "Bond Question" in race_string or "Retain" in race_string:
        return ""
    elif "(REP)" in race_string or "REPUBLICAN" in race_string:
        return "R"
    elif "(DEM)" in race_string or "(Democrat)" in race_string or "DEMOCRAT" in race_string:
        return "D"
    elif "(L)" in race_string or "(LBT)" in race_string or "LIBERTARIAN" in race_string:
        return "L"
    elif "(WI)" in race_string:
        return "O"
    elif "(IND)" in race_string:
        return "I"
    elif "(NOPTY)" in race_string:
        return "N"
    elif race_string[0:3]=="YES":
        return "YES"
    elif race_string[0:2]=="NO":
        return "NO"
    else:
        return race_string.split("-:-")[-1][0]
           
def get_name(name_string):
    if "No." in name_string:
        return ""
    else:
        #print(name_string)
        name_string = name_string.replace(" (write in)","")
        name_string = name_string.split("-:-")[1]
        name_string = name_string.replace("'","")
        likely_last = name_string.split(" ")[-1]
        proposed_last = likely_last[:3]
        if proposed_last in ['II', 'III', 'Jr', 'Jr.', 'Sr.', 'JR.', "JR", "IV", 'I']:
            likely_last = name_string.split(" ")[-2]
            proposed_last = likely_last[:3]
        #print(proposed_last.upper())
        return proposed_last.upper()
#     else:
#         return name_string.split(" ")[-1][:3].upper()
import re
 

def get_district(race_string, fill_level):
    race_string = race_string.replace('"',"'")
    temp = race_string
    return re.sub("\D","",temp).zfill(fill_level)

def column_rename_function(name_string):
    election_type = get_election_type(name_string)
    year = "24"
    party = get_party(name_string)
    race = get_race(name_string)
    district = ""
    #print(name_string)
    if race in ["CON", "SU"]:
        district = get_district(name_string, 2)
        year = ""
    elif race in ["SL"]:
        district = get_district(name_string, 3)
        year = ""
    elif race in ["SC"]:
        district = get_district(name_string, 1)
        year = ""
    name = get_name(name_string)
    if "BI" in race:
        party = ""
    new_col_name = election_type + year + race + district + party + name
    print(name_string)
    print("election_type: ", election_type)
    print("year: ", year)
    print("race: ", race)
    print("district: ", district)
    print("party: ", party)
    print("name: ", name)
    if len(new_col_name) > 10:
        print("LONG NAME")
        print(name_string, "->", new_col_name)
    return new_col_name

# Make a dictionary that points to the new column names and checks for duplicates

#la_20_state_level.rename(columns = {"Lessie Olivia Leblanc (DEM)-:-U. S. Representative -- 3rd Congressional District":"Lessie OliviaLeblanc (DEM)-:-U. S. Representative -- 3rd Congressional District"}, inplace = True)

races_to_clean = [i for i in pivoted_filtered.columns if i not in ["pivot_col","JurisName"]]


race_updates_dict = {}
race_updates_reversed = {}
#clean_dups = {}
new_names = []
for val in races_to_clean:
    if "Judge" not in val:
        new_name = column_rename_function(val)
        race_updates_dict[val] = new_name
        if new_name not in new_names:
            new_names.append(new_name)
            race_updates_reversed[new_name] = val
        else:
            print("Duplicate", new_name)
            print(race_updates_reversed[new_name])
            print(val)
            #clean_dups[val] = race_updates_reversed[new_name]
    else:
        print("ADD:", val)
```

```python
def get_race(race_string):
    race_string = race_string.title()
    race_string = race_string.replace("(Vote For 1)","")
    if "U.S. House" in race_string or 'Us House' in race_string or "Representative To Congress" in race_string or "U. S. Representative" in race_string or "United States Representative" in race_string or "Congress" in race_string:
        return "CON"
    elif "State House" in race_string or "State Representative" in race_string or "Representative" in race_string:
        return "SL"
    elif "State Senate" in race_string or "State Senator" in race_string or "Senate" in race_string:
        return "SU"
    elif "President" in race_string:
        return "PRE"
    elif "US Senate" in race_string or "Us Senate" in race_string or "U.S. Senator" in race_string or "U. S. Senator" in race_string or "United States Senator" in race_string:
        return "USS"
    elif "Public Service" in race_string:
        return "PSC"
    elif "Attorney General" in race_string:
        return "ATG"
    elif "Auditor General" in race_string or "Auditor" in race_string:
        return "AUD"
    elif "Treasurer" in race_string:
        return "TRE"
    elif "Superintendent" in race_string:
        return "SUP"
    elif "Secretary Of State" in race_string:
        return "SOS"
    elif "Governor" in race_string:
        return "GOV"
    elif "Lieutenant Governor" in race_string:
        return "LTG"

    elif "Commissioner Of Labor" in race_string:
        return "LAB"
    elif "Commissioner Of Agriculture" in race_string:
        return "AGR"
    elif "Commissioner Of Insurance" in race_string:
        return "INS"
    elif "Chief Justice Of The Supreme Court" in race_string:
        return "CJU"
    elif "Supreme" in race_string:
        return "SC"
    elif "Justice Of The Supreme Court" in race_string:
        return "JUS"
    elif "Amendment" in race_string:
        #num = race_string.split(" ")[2][:-1]
        return "AMD"
    elif "Bond Question" in race_string:
        num = race_string.split(" ")[2][:-1]
        return "BQ"+str(num)
    elif "Judicial Retention" in race_string:
        return "JR"
    elif "Commissioner Of Public Lands" in race_string:
        return "CPL"
    elif "Comptroller" in race_string:
        return "COM"
    
    elif "All Medically Appropriate Assisted" in race_string:
        return "BI1"
    
    elif "Should Any Candidate Appearing On The Illinois Ballot For Federal" in race_string:
        return "BI3"
    
    elif "Illinois Constitution Be Amended To Create An Additional 3%" in race_string:
        return "BI2"

    else:
        print("No race for:", race_string)
        raise ValueError
        
def get_election_type(race_string):
    return "G"
        
def get_party(race_string):
    race_string = race_string.replace('"',"'")

    if "JUDICIAL RETENTION" in race_string or "Amendment" in race_string or "amendment" in race_string or "Bond Question" in race_string or "Retain" in race_string:
        return ""
    elif "(REP)" in race_string or "REPUBLICAN" in race_string:
        return "R"
    elif "(DEM)" in race_string or "(Democrat)" in race_string or "DEMOCRAT" in race_string:
        return "D"
    elif "(L)" in race_string or "(LBT)" in race_string or "LIBERTARIAN" in race_string:
        return "L"
    elif "(WI)" in race_string:
        return "O"
    elif "(IND)" in race_string:
        return "I"
    elif "(NOPTY)" in race_string:
        return "N"
    elif race_string[0:3]=="YES":
        return "YES"
    elif race_string[0:2]=="NO":
        return "NO"
    else:
        return race_string.split("-:-")[-1][0]
           
def get_name(name_string):
    if "No." in name_string:
        return ""
    else:
        #print(name_string)
        name_string = name_string.replace(" (write in)","")
        name_string = name_string.split("-:-")[1]
        name_string = name_string.replace("'","")
        likely_last = name_string.split(" ")[-1]
        proposed_last = likely_last[:3]
        if proposed_last in ['II', 'III', 'Jr', 'Jr.', 'Sr.', 'JR.', "JR", "IV", 'I']:
            likely_last = name_string.split(" ")[-2]
            proposed_last = likely_last[:3]
        #print(proposed_last.upper())
        return proposed_last.upper()
#     else:
#         return name_string.split(" ")[-1][:3].upper()
import re
 

def get_district(race_string, fill_level):
    race_string = race_string.replace('"',"'")
    temp = race_string
    return re.sub("\D","",temp).zfill(fill_level)

def column_rename_function(name_string):
    election_type = get_election_type(name_string)
    year = "24"
    party = get_party(name_string)
    race = get_race(name_string)
    district = ""
    #print(name_string)
    if race in ["CON", "SU"]:
        district = get_district(name_string, 2)
        year = ""
    elif race in ["SL"]:
        district = get_district(name_string, 3)
        year = ""
    elif race in ["SC"]:
        district = get_district(name_string, 1)
        year = ""
    name = get_name(name_string)
    if "BI" in race:
        party = ""
    new_col_name = election_type + year + race + district + party + name
    print(name_string)
    print("election_type: ", election_type)
    print("year: ", year)
    print("race: ", race)
    print("district: ", district)
    print("party: ", party)
    print("name: ", name)
    if len(new_col_name) > 10:
        print("LONG NAME")
        print(name_string, "->", new_col_name)
    return new_col_name

# Make a dictionary that points to the new column names and checks for duplicates

#la_20_state_level.rename(columns = {"Lessie Olivia Leblanc (DEM)-:-U. S. Representative -- 3rd Congressional District":"Lessie OliviaLeblanc (DEM)-:-U. S. Representative -- 3rd Congressional District"}, inplace = True)

races_to_clean = [i for i in pivoted_county.columns if i not in ["County"]]


race_updates_dict = {}
race_updates_reversed = {}
#clean_dups = {}
new_names = []
for val in races_to_clean:
    if "Judge" not in val:
        new_name = column_rename_function(val)
        race_updates_dict[val] = new_name
        if new_name not in new_names:
            new_names.append(new_name)
            race_updates_reversed[new_name] = val
        else:
            print("Duplicate", new_name)
            print(race_updates_reversed[new_name])
            print(val)
            #clean_dups[val] = race_updates_reversed[new_name]
    else:
        print("ADD:", val)
```
