```python
def get_race(race_string, print_debug):
    race_string = race_string.title()
    race_string = race_string.replace("(Vote For 1)","")
    if "U.S. House" in race_string or 'Us House' in race_string or "Representative To Congress" in race_string or "U. S. Representative" in race_string or "United States Representative" in race_string or "Congress" in race_string or "U S House" in race_string or "United States House Of Representatives" in race_string:
        return "CON"
    elif "State House" in race_string or "State Representative" in race_string or "Representative" in race_string or "House Of Delegates" in race_string:
        return "SL"
    elif "State Senate" in race_string or "State Senator" in race_string or "Senate Of Virginia" in race_string or "Tennessee Senate District" in race_string:
        return "SU"
    elif "President" in race_string:
        return "PRE"
    elif "US Senate" in race_string or "Us Senate" in race_string or "U.S. Senator" in race_string or "U. S. Senator" in race_string or "United States Senator" in race_string or "U S Senate" in race_string or "United States Senate" in race_string:
        return "USS"
    elif "Public Service" in race_string or "Psc" in race_string:
        return "PSC"+re.sub("\D","",race_string)
    elif "Attorney General" in race_string:
        return "ATG"
    elif "Auditor General" in race_string or "Auditor" in race_string:
        return "AUD"
    elif "Treasurer" in race_string:
        return "TRE"
    elif "Superintendent" in race_string:
        return "SUP"
    elif "Secretary Of State" in race_string:
        return "SOS"
    elif "Lieutenant Governor" in race_string:
        return "LTG"
    elif "Governor" in race_string:
        return "GOV"


    elif "Commissioner Of Labor" in race_string:
        return "LAB"
    elif "Commissioner Of Agriculture" in race_string:
        return "AGR"
    elif "Commissioner Of Insurance" in race_string:
        return "INS"
    elif "Chief Justice Of The Supreme Court" in race_string:
        return "CJU"
    elif "Supreme" in race_string:
        return "SC"
    elif "Justice Of The Supreme Court" in race_string:
        return "JUS"
    elif "Party Question" in race_string:
        return "PQ"+re.sub("\D","",race_string)
    elif "Amendment" in race_string:
        num = race_string.split(" ")[2]

        return "AMD"+num
    elif "Bond Question" in race_string:
        num = race_string.split(" ")[3]
        return "BQ"+str(num)
    elif "Judicial Retention" in race_string:
        return "JR"
    elif "Commissioner Of Public Lands" in race_string:
        return "CPL"
    elif "Comptroller" in race_string:
        return "COM"
    
    elif "All Medically Appropriate Assisted" in race_string:
        return "BI1"
    
    elif "Should Any Candidate Appearing On The Illinois Ballot For Federal" in race_string:
        return "BI3"
    
    elif "Illinois Constitution Be Amended To Create An Additional 3%" in race_string:
        return "BI2"

    else:
        if print_debug:
            print("No race for:", race_string)
        raise ValueError
        
def get_election_type(race_string, print_debug):
    if "Special" in race_string:
        return "S"
    else:
        return "R"
        
def get_party(race_string, print_debug):
    race_string = race_string.split("-:-")[1]
    race_string = race_string.replace('"',"'")

    if "JUDICIAL RETENTION" in race_string or "Amendment" in race_string or "amendment" in race_string or "Bond Question" in race_string or "Retain" in race_string:
        return ""
    elif "(REP)" in race_string or "REPUBLICAN" in race_string:
        return "R"
    elif "(DEM)" in race_string or "(Democrat)" in race_string or "DEMOCRAT" in race_string:
        return "D"
    elif "(L)" in race_string or "(LBT)" in race_string or "LIBERTARIAN" in race_string:
        return "L"
    elif "(WI)" in race_string:
        return "O"
    elif "(IND)" in race_string:
        return "I"
    elif "(NOPTY)" in race_string:
        return "N"
    elif race_string[0:3]=="YES":
        return "YES"
    elif "NO PARTY" in race_string:
        return "O"
    else:
        return race_string[0]
           
def get_name(name_string, print_debug):
    name_string_unmodified = name_string
    if "No." in name_string:
        return ""
    elif "All Others" in name_string:
        return "OTH"
    else:
        #print(name_string)
        name_string = name_string.replace(" (write in)","").replace(" (I)","")
        if "President" in name_string:
            name_string = name_string.split("/")[0]        
        else:
            name_string = name_string.split("-:-")[2]
            
        if name_string == "Sharon Stokes Williamson":
            return "WLM"

            
        name_string = name_string.replace("'","")
        name_string = name_string.split(" / ")[0]
        likely_last = name_string.split(" ")[-1]
        if "Special State Senate" in name_string_unmodified:
            likely_last = name_string.split(" ")[-2]
        proposed_last = likely_last[:3]
        if proposed_last in ['II', 'III', 'Jr', 'Jr.', 'Sr.', 'JR.', "JR", "IV", 'I']:
            likely_last = name_string.split(" ")[-2]
            proposed_last = likely_last[:3]
        return proposed_last.upper()
#     else:
#         return name_string.split(" ")[-1][:3].upper()
import re
 

def get_district(race_string, fill_level, print_debug):
    race_string = race_string.replace('"',"'").replace(" (Vote for 1)","").replace(" including precincts 1","")
    temp = race_string
    if "State Representative" in race_string:
        if "Berkshire" in race_string:
            return "BK"+re.sub("\D","",temp)
        elif "Barnstable" in race_string:
            if "Barnstable Dukes and Nantucket" in race_string:
                return "BDN"
            return "BN"+re.sub("\D","",temp)
        elif "Hampshire" in race_string:
            return "HS"+re.sub("\D","",temp)
        return race_string.split("-:-")[0].split(" ")[-1].zfill(fill_level)
    elif "State Senate" in race_string:
        if "Third Bristol Plymouth District" in race_string:
            return "3BP"
        elif "-Worcester Hampden District" in race_string:
            return "WHM"
        elif "-Hampden District" in race_string:
            return "HMN"
        elif "Plymouth and Barnstable District" in race_string:
            return "PLB"
        elif "2nd Essex and Middlesex District" in race_string:
            return "2EM"
        elif "1st Essex District" in race_string:
            return "1ED"
        elif "Norfolk and Suffolk District" in race_string:
            return "NRS"
        elif "Middlesex and Norfolk District" in race_string:
            return "MXN"
        elif "Middlesex and Suffolk District" in race_string:
            return "MXS"
        elif "Bristol and Norfolk District" in race_string:
            return "BLN"
        elif "Norfolk Worcester Middlesex District" in race_string:
            return "NWM"
        elif "5th Middlesex District" in race_string:
            return "5MX"
        elif "4th Middlesex District" in race_string:
            return "4MX"
        elif "3rd Middlesex District" in race_string:
            return "3MX"
        elif "2nd Middlesex District" in race_string:
            return "2MX"
        elif "1st Middlesex District" in race_string:
            return "1MX"
        elif "Second Plymouth Norfolk District" in race_string:
            return "2PN"
        elif "Worcester and Middlesex District" in race_string:
            return "WRM"
        elif "1st Bristol and Plymouth District" in race_string:
            return "1BP"

        elif "Hampshire Franklin and Worcester District" in race_string:
            return "HFW"
        elif "Berkshire Hampden Franklin Hampshire District" in race_string:
            return "BHF"
        elif "1st Worcester District" in race_string:
            return "1WR"
        elif "Hampden Hampshire Worcester District" in race_string:
            return "HHW"
        elif "Cape and Islands District" in race_string:
            return "CPI"
        elif "Suffolk and Middlesex District" in race_string:
            return "SKM"
        elif "2nd Essex District" in race_string:
            return "2ES"
        elif "3rd Essex District" in race_string:
            return "3ES"
        elif "Norfolk Middlesex District" in race_string:
            return "NKM"
        elif "1st Essex and Middlesex District" in race_string:
            return "1EM"
        elif "2nd Suffolk District" in race_string:
            return "2SF"
        elif "1st Suffolk District" in race_string:
            return "1SF"
        elif "2nd Bristol and Plymouth District" in race_string:
            return "2BP"
        elif "Norfolk Plymouth Bristol District" in race_string:
            return "NPB"
        elif "Middlesex and Worcester District" in race_string:
            return "MXW"
        elif "Worcester Hampshire District" in race_string:
            return "WRH"
        elif "3rd Suffolk District" in race_string:
            return "3SF"
        elif "Norfolk and Plymouth District" in race_string:
            return "NRP"
        elif "First Plymouth Norfolk District" in race_string:
            return "1PN"
        elif "2nd Worcester District" in race_string:
            return "2WR"
        elif "Hampden and Hampshire District" in race_string:
            return "HMH"
        else:
            return re.sub("\D","",temp).zfill(fill_level)
        

#         else:
#             print("FIX", race_string)
#             raise ValueError
                
        
    else:
        return re.sub("\D","",temp).zfill(fill_level)

def column_rename_function(name_string, print_debug):
    election_type = get_election_type(name_string, print_debug)
    year = "26"
    party = get_party(name_string,print_debug)
    race = get_race(name_string,print_debug)
    district = ""
    #print(name_string)
    if race in ["CON", "SU"]:
        district = get_district(name_string, 2,print_debug)
        year = ""
    elif race in ["SL"]:
        district = get_district(name_string, 3,print_debug)
        year = ""
    elif race in ["SC"]:
        district = get_district(name_string, 1,print_debug)
        year = ""
    elif "PSC" in race or "PQ" in race:
        year = ""
    name = get_name(name_string,print_debug)
    if "AMD" in race:
        party = ""
    if "BI" in race:
        party = ""
    new_col_name = election_type + year + race + district + party + name
    if print_debug:
        print(name_string)
        print("election_type: ", election_type)
        print("year: ", year)
        print("race: ", race)
        print("district: ", district)
        print("party: ", party)
        print("name: ", name)
    if len(new_col_name) > 10:
        print("LONG NAME")
        print(name_string, "->", new_col_name)
        raise ValueError
    return new_col_name

# Make a dictionary that points to the new column names and checks for duplicates


def get_column_renaming_vals(df, non_contest_names_list, print_debug = False):

    races_to_clean = [i for i in df.columns if i not in non_contest_names_list]


    race_updates_dict = {}
    race_updates_reversed = {}
    #clean_dups = {}
    new_names = []
    error_w_rename = []
    duplicate_contest = []
    skipped_contest = []
    for val in races_to_clean:
        if "Clerk of Courts" not in val and "Register of Deeds" not in val and "Governor s Council" not in val:
            try:
                new_name = column_rename_function(val, print_debug)
                race_updates_dict[val] = new_name
                if new_name not in new_names:
                    new_names.append(new_name)
                    race_updates_reversed[new_name] = val
                else:
                    duplicate_contest.append(new_name)

                    if print_debug:
                        print("Duplicate", new_name)
                        print(race_updates_reversed[new_name])
                        print(val)
                    #clean_dups[val] = race_updates_reversed[new_name]
            except ValueError as v:
                error_w_rename.append(val)
        else:
            skipped_contest.append(val)
            
    if len(error_w_rename)>0:
        print("Error Renaming")
        print(error_w_rename)
        
    if len(duplicate_contest)>0:
        print("Duplicate Contest")
        print(duplicate_contest)
        
    if len(skipped_contest) > 0:
        print("Skipped Contest")
        print(skipped_contest)        
            
    return race_updates_dict, race_updates_reversed, error_w_rename, duplicate_contest, skipped_contest
```

```python
race_updates_dict, race_updates_reversed, error_w_rename, duplicate_contest, skipped_contest = get_column_renaming_vals(data, ['Prec_ID'], print_debug = False)
```

```python
data.rename(columns = race_updates_dict, inplace = True)
```
